```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/apple.webp')
```

```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
```

```python
image_float = gray_image.astype(np.float32)
```

```python
addition = np.clip(image_float + 50, 0, 255).astype(np.uint8)

subtraction = np.clip(image_float - 50, 0, 255).astype(np.uint8)

multiplication = np.clip(
    image_float * 1.5, 0, 255
).astype(np.uint8)

division = np.clip(
    image_float / 2, 0, 255
).astype(np.uint8)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(addition, cmap='gray')
plt.title('Addition (+50)')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(subtraction, cmap='gray')
plt.title('Subtraction (-50)')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(multiplication, cmap='gray')
plt.title('Multiplication (×1.5)')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(division, cmap='gray')
plt.title('Division (÷2)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
