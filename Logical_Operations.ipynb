```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/apple.webp')
```

```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
```

```python
_, binary1 = cv2.threshold(
    gray_image, 100, 255, cv2.THRESH_BINARY
)

_, binary2 = cv2.threshold(
    gray_image, 160, 255, cv2.THRESH_BINARY
)
```

```python
and_image = cv2.bitwise_and(binary1, binary2)
or_image = cv2.bitwise_or(binary1, binary2)
not_image = cv2.bitwise_not(binary1)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(binary1, cmap='gray')
plt.title('Binary Image 1')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(binary2, cmap='gray')
plt.title('Binary Image 2')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(and_image, cmap='gray')
plt.title('AND Operation')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(or_image, cmap='gray')
plt.title('OR Operation')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(not_image, cmap='gray')
plt.title('NOT Operation')
plt.axis('off')

plt.tight_layout()
plt.show()
```
